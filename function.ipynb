```python
# keyword argument
def  my_function(*,name):
    print("hello", name)
my_function(name= "emil")
```

```python
def my_function(a, b, /, *, c, d):
    return a + b + c + d

result = my_function(10, 20, c=30, d=40)
print(result)
```
