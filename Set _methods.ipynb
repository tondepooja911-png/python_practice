# union

```python
s1 = {1,2,3,4}
s2 = {3,4,5,6}
print(s1.union(s2))
```

# difference

```python
p1 = {1,2,3,4,5}
p2 = {1,4,7,8,9}
p2.difference(p1)
```

 # intersection

```python
a ={10,20,30,40}
b = {50,60,70,80}
a.intersection(b)
```

```python
c = {1,2,3,4,5}
d = {1,2,4,5,9}
c.intersection(d)
```
